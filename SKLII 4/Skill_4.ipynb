```python

import os

project = "/content/pytorch-cuda-project"

os.makedirs(f"{project}/src", exist_ok=True)
os.makedirs(f"{project}/tests", exist_ok=True)

os.chdir(project)

with open("README.md", "w") as f:
    f.write("# PyTorch CUDA Project\n\nProject repository and Git workflow.")

with open("requirements.txt", "w") as f:
    f.write("torch\n")

with open(".gitignore", "w") as f:
    f.write("__pycache__/\n*.pyc\n.env\n")

print("Project repository created at:", project)
```

```python

code = '''
import torch

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

a = torch.randn(3, 3, device=device)
b = torch.randn(3, 3, device=device)

print("Device:", device)
print("Matrix multiplication result:")
print(torch.matmul(a, b))
'''

with open("src/main.py", "w") as f:
    f.write(code)

exec(compile(code, "src/main.py", "exec"))
```

```python

import subprocess

def git(*args):
    result = subprocess.run(
        ["git", *args],
        capture_output=True,
        text=True
    )
    print(result.stdout or result.stderr)

git("init")
git("config", "user.name", "Colab Student")
git("config", "user.email", "student@example.com")
git("add", ".")
git("commit", "-m", "Initial project setup")
git("status")
git("log", "--oneline")
```

```python

import os

project = "/content/pytorch-cuda-project"

os.makedirs(f"{project}/src", exist_ok=True)
os.makedirs(f"{project}/tests", exist_ok=True)

files = {
    "README.md": "# PyTorch CUDA Project\n",
    "requirements.txt": "torch\n",
    ".gitignore": "__pycache__/\n*.pyc\n.env\n",
    "src/main.py": "import torch\nprint('PyTorch:', torch.__version__)\nprint('CUDA available:', torch.cuda.is_available())\n"
}

for name, content in files.items():
    with open(os.path.join(project, name), "w") as f:
        f.write(content)

print("Project created successfully!")
print(os.listdir(project))
```
